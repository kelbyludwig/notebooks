```python
# https://doc.sagemath.org/html/en/reference/plotting/sage/plot/bar_chart.html
from sage.plot.bar_chart import bar_chart
bar_chart([x^2 for x in range(1,20)], width=0.2)
```

```python
# https://doc.sagemath.org/html/en/reference/plotting/sage/plot/histogram.html
from sage.plot.histogram import histogram
histogram([x^2 for x in range(1,20)], bins=10)
```

```python
# https://doc.sagemath.org/html/en/reference/plotting/sage/graphs/graph_plot.html
D = DiGraph({'k':['k',1], 1:[2], 2:[3]}, loops=True)
D.show()
```

```python
# https://doc.sagemath.org/html/en/reference/plotting/sage/plot/scatter_plot.html
from sage.plot.scatter_plot import scatter_plot
scatter_plot([(0, 0), (.5, 50), (1, 1)], markersize=100, facecolor='red', ymax=100)
```

```python
# https://doc.sagemath.org/html/en/reference/plotting/sage/plot/density_plot.html
from sage.plot.density_plot import density_plot

x,y = var('x,y')
density_plot(sin(x) * sin(y), (x,-2,2), (y,-2,2))
```
