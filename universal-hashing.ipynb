````python
def get_lcg(a, b, p):
    assert(a != 0)
    assert(a > 0 and a < p)
    assert(b > 0 and b < p)
    # assert(is_prime(p))
    def h(x):
        return (a*x + b) % p
    return h

# sane(?) parameter selection: gcd(a, p) = 1
func = get_lcg(a=5, b=4, p=11)
collisions = {}
for i in range(11):
    l = collisions.get(func(i), [])
    l.append(i)
    collisions[func(i)] = l
print(collisions)

# adversarial parameter selection: gcd(a, p) != 1
func = get_lcg(a=3, b=4, p=9)
collisions = {}
for i in range(9):
    l = collisions.get(func(i), [])
    l.append(i)
    collisions[func(i)] = l
print(collisions)
````

## Collisions in mod p

A collision in `(a*x + b) % p` implies a collision `mod m`. A non-collision `(a*x + b) % p` *does not* imply no collision `mod m`.

## Are collisions mod p possible for fixed parameters?

Given restrictions on the range of input, no collisions can occur for two distinct `x`, `x'` in the function `f(x) = a*x + b % p` for fixed `a`,`b`,`p`. Proof:

Assume `0 <= x < p` and `0 <= x' < p` and `x != x'`.

A collision implies their `f` outputs are equal. Assuming there is a collision shows a contradiction:
```
a*x + b = a*x' + b (mod p)
a*x = a*x' (mod p) # substracted b
x = x' (mod p)     # multiplied by a^-1 (NOTE THIS MUST MEAN a IS INVERTIBLE)
```

If `a` is invertible and `x` and `x'` are non-equal and both are the least positive residue `mod p` this is a contradiction.

## When you add mod m what is the probability of a collision?

Given randomly selected `a`, `b`, a fixed `p` and `m`, and two inputs `x`, `x'` a collision happens when:
```
ax + b = ax' + b (mod p) (mod m)
ax + b = ax' + b (mod m) (mod p)
ax + b = ax' + b - im (mod p)
ax = ax' - im (mod p)
ax - ax' = -im (mod p)
a(x - x') = -im (mod p)
a(x - x') + im = 0 (mod p)
a(x - x') = 0 (mod m) (mod p)
```
